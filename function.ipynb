```python
lst = [ x for x in range(500) ] # dataset

def data_loader(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]


x = data_loader(5,lst)

print(next(x))
print(next(x))
```

```python
lt = [x for x in range(500) ] 

def data_load(chunk_size,lst): 
    for i in range(0,len(lst),chunk_size):
        yield lst[i : i + chunk_size]

        
x = data_load(5,lt)

print(next(x))
print(next(x))
```

```python
max_num = lambda a, b: a if a > b else b

print(max_num(12, 25))
```
